| Data Preparation | Zulässige Merkmale, Split-Logik, Kodierung/Skalierung/Imputation und Leakage-Prüfung. |
| Modeling | Baseline sowie mindestens zwei begründete Ansätze oder Konfigurationen. |
| Evaluation | Cross-Validation, finale Testbewertung bzw. Clustergüte und interpretierte Fehlermuster. |
| Interpretation | Grenzen, Risiken, fachliche Konsequenz und nächster Schritt. |
